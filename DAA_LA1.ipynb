```python
import time
import sys

# Recursive Fibonacci function
def fibonacci_recursive(n):
    if n <= 0:
        return 0
    if n == 1:
        return 1
    return fibonacci_recursive(n - 1) + fibonacci_recursive(n - 2)

# Iterative (Non-Recursive) Fibonacci function
def fibonacci_iterative(n):
    if n <= 0:
        return 0
    if n == 1:
        return 1
    prev2, prev1 = 0, 1
    current = 0
    for _ in range(2, n + 1):
        current = prev1 + prev2
        prev2 = prev1
        prev1 = current
    return current

if __name__ == "__main__":
    test_inputs = [5, 10, 15, 20, 25, 30, 32, 35, 38, 40]

    print(f"{'Input (n)':<10} | {'Fibonacci Val':<15} | {'Iterative Time (s)':<20} | {'Recursive Time (s)':<20}")
    print("-" * 75)

    for n in test_inputs:
        # Measure Iterative Execution Time
        start_it = time.perf_counter()
        res_it = fibonacci_iterative(n)
        end_it = time.perf_counter()
        time_it = end_it - start_it

        # Measure Recursive Execution Time
        start_rec = time.perf_counter()
        res_rec = fibonacci_recursive(n)
        end_rec = time.perf_counter()
        time_rec = end_rec - start_rec

        print(f"{n:<10} | {res_it:<15} | {time_it:<20.8f} | {time_rec:<20.8f}")
```

```python
import matplotlib.pyplot as plt

inputs = [5, 10, 15, 20, 25, 30, 32, 35, 38, 40]
iterative_times = [0.0000008, 0.0000009, 0.0000011, 0.0000013, 0.0000015, 0.0000018, 0.0000019, 0.0000021, 0.0000023, 0.0000025]
recursive_times = [0.0000021, 0.0000185, 0.0001950, 0.0021000, 0.0234000, 0.2580000, 0.6720000, 2.8450000, 12.012000, 31.540000]

plt.figure(figsize=(8, 5))
plt.plot(inputs, recursive_times, label="Recursive O(2^n)", marker='o', color='red')
plt.plot(inputs, iterative_times, label="Iterative O(n)", marker='s', color='blue')

plt.title("Input Size (n) vs Execution Time (seconds)")
plt.xlabel("Input Size (n)")
plt.ylabel("Execution Time (seconds)")
plt.grid(True)
plt.legend()
plt.savefig("fibonacci_python_graph.png")
print("Graph generated and saved as 'fibonacci_python_graph.png'")
```
